```python
class calculator():
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b
```

```python
cal=calculator(2,3)
```

```python
def calmean(listvals):
    return sum(listvals)/len(listvals)

def calmedian(listvals):
    sorted_list = sorted(listvals)
    n = len(sorted_list)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_list[mid - 1] + sorted_list[mid]) / 2
    else:
        return sorted_list[mid]

def calstddev(listvals):
    mean = calmean(listvals)
    variance = sum((x - mean) ** 2 for x in listvals) / len(listvals)
    return variance ** 0.5
```

```python
class StatsCalculator:
    def __init__(self, data):
        self.data = data

    def mean(self):
        return calmean(self.data)

    def median(self):
        return calmedian(self.data)

    def stddev(self):
        return calstddev(self.data)
```

```python
listvals = [11, 2, 23, 41, 5]
st = StatsCalculator(listvals)
mean_value = st.mean()
median_value = st.median()
stddev_value = st.stddev()
print(f"Mean: {mean_value}, Median: {median_value}, Standard Deviation: {stddev_value}")
```
